```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

print("Libraries imported successfully!")
```

```python
df = pd.read_csv("air_quality_daily.csv")

print("Dataset loaded successfully!")

display(df.head())
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe())
```

```python
print(df.isnull().sum())
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
numeric_cols = df.select_dtypes(include=np.number).columns

df[numeric_cols].hist(
    figsize=(15, 10),
    bins=20,
    edgecolor="black"
)

plt.suptitle("Distribution of Air Quality Parameters", fontsize=16)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(14, 6))

sns.boxplot(data=df[numeric_cols])

plt.title("Air Quality Parameters - Outlier Analysis")
plt.xlabel("Parameters")
plt.ylabel("Values")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 8))

correlation = df[numeric_cols].corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Between Air Quality Parameters")
plt.tight_layout()
plt.show()
```

```python
if len(numeric_cols) >= 2:

    x_col = numeric_cols[0]
    y_col = numeric_cols[1]

    plt.figure(figsize=(10, 6))

    sns.scatterplot(
        data=df,
        x=x_col,
        y=y_col
    )

    plt.title(f"Relationship Between {x_col} and {y_col}")
    plt.xlabel(x_col)
    plt.ylabel(y_col)

    plt.tight_layout()
    plt.show()

else:
    print("Not enough numerical columns.")
```

```python
categorical_cols = df.select_dtypes(
    include=["object", "category", "str"]
).columns

if len(categorical_cols) > 0:

    col = categorical_cols[0]

    plt.figure(figsize=(10, 6))

    df[col].value_counts().head(10).plot(
        kind="bar",
        edgecolor="black"
    )

    plt.title(f"Top 10 Categories in {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.xticks(rotation=45)

    plt.tight_layout()
    plt.show()

else:
    print("No categorical columns found.")
```

```python
fig, axes = plt.subplots(2, 2, figsize=(16, 10))

# 1. Histogram
if len(numeric_cols) > 0:
    df[numeric_cols[0]].dropna().hist(
        ax=axes[0, 0],
        bins=20,
        edgecolor="black"
    )
    axes[0, 0].set_title("Data Distribution")

# 2. Boxplot
if len(numeric_cols) > 0:
    sns.boxplot(
        y=df[numeric_cols[0]],
        ax=axes[0, 1]
    )
    axes[0, 1].set_title("Outlier Analysis")

# 3. Correlation Heatmap
if len(numeric_cols) >= 2:
    sns.heatmap(
        df[numeric_cols].corr(),
        cmap="coolwarm",
        ax=axes[1, 0]
    )
    axes[1, 0].set_title("Correlation Analysis")

# 4. Bar Chart
if len(categorical_cols) > 0:
    df[categorical_cols[0]].value_counts().head(10).plot(
        kind="bar",
        ax=axes[1, 1]
    )
    axes[1, 1].set_title("Category Analysis")
    axes[1, 1].tick_params(axis="x", rotation=45)

plt.suptitle("AIR QUALITY DATA VISUALIZATION DASHBOARD",
             fontsize=18)

plt.tight_layout()
plt.show()
```
