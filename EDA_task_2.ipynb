```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("air_quality_daily.csv")
df.head()
```

```python
print("First Five Rows:")
print(df.head())

print("\nLast Five Rows:")
print(df.tail())

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
print(df.describe(include="all"))
```

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nTotal Missing Values:")
print(df.isnull().sum().sum())
```

```python
print("Duplicate Records:")
print(df.duplicated().sum())

df = df.drop_duplicates()

print("\nDuplicates Removed Successfully")
```

```python
print("Numerical Data Analysis:")
print(df.describe())

print("\nUnique Values in Each Column:")
for col in df.columns:
    print(col, ":", df[col].nunique())
```

```python
numeric_df = df.select_dtypes(include=np.number)

print("Correlation Between Numerical Variables:")
print(numeric_df.corr())
```

```python
numeric_df.hist(figsize=(12, 8), bins=15)

plt.suptitle("Distribution of Numerical Variables")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.heatmap(numeric_df.corr(), annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap")
plt.show()
```

```python
for col in numeric_df.columns:
    plt.figure(figsize=(7, 4))
    sns.boxplot(x=df[col])
    plt.title("Outlier Detection: " + col)
    plt.show()
```

```python
if len(numeric_df.columns) >= 2:

    x = numeric_df.columns[0]
    y = numeric_df.columns[1]

    plt.figure(figsize=(7, 5))

    sns.scatterplot(x=df[x], y=df[y])

    plt.title(x + " vs " + y)
    plt.show()
```

```python
for col in numeric_df.columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print("\nOutliers in", col, ":", len(outliers))
```

```python
print("FINAL DATA QUALITY REPORT")

print("\nNumber of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())

print("\nData Types:")
print(df.dtypes)

print("\nEDA Completed Successfully!")
```

```python
# Identify categorical columns
categorical_columns = df.select_dtypes(include=['object', 'category']).columns

print("Categorical Columns:")
print(categorical_columns)

# Analyse each categorical column
for col in categorical_columns:
    print("\nColumn:", col)
    print(df[col].value_counts())

    print("\nPercentage:")
    print(df[col].value_counts(normalize=True) * 100)
```

```python
# Check inconsistent text values
for col in categorical_columns:

    print("\nColumn:", col)

    print(df[col].dropna().astype(str).unique())
```

```python
for col in categorical_columns:

    df[col] = df[col].apply(
        lambda x: x.strip().lower() if isinstance(x, str) else x
    )

print("Text values standardized successfully")
```

```python
print("Before Conversion:")
print(df.dtypes)

# Try converting numeric-looking text columns
for col in df.select_dtypes(include='str').columns:

    converted = pd.to_numeric(df[col], errors='coerce')

    if converted.notna().sum() > 0:
        if converted.notna().sum() == df[col].notna().sum():
            df[col] = converted

print("\nAfter Conversion:")
print(df.dtypes)
```

```python
for col in numeric_df.columns:
    plt.figure(figsize=(7, 4))
    sns.histplot(df[col].dropna(), kde=True)
    plt.title("Distribution of " + col)
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.heatmap(
    df.select_dtypes(include=np.number).corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
print("FINAL EDA FINDINGS")

print("\nDataset Shape:")
print(df.shape)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())

print("\nNumerical Summary:")
print(df.describe())

print("\nCategorical Summary:")
print(df.describe(include='str'))

print("\nEDA Completed Successfully!")
```
